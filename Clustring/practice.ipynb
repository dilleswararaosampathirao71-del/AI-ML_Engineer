```python
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.datasets import load_iris
```

```python
df=load_iris()
x=df.data
y=df.target
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x=scaler.fit_transform(x)
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
pcaData=pca.fit_transform(x)
```

```python
from kneed import KneeLocator
wcss=[]
for k in range(1,21):
    model=KMeans(n_clusters=k)
    labels=model.fit_predict(pcaData)
    wcss.append(model.inertia_)  
knee=KneeLocator(range(1,21),wcss,curve="convex",direction="decreasing")
k=knee.knee
print(k)    
```

```python
ss=[]
for k in range(2,21):
    model=KMeans(n_clusters=k)
    labels=model.fit_predict(pcaData)
    score=silhouette_score(x,labels)
    ss.append(score)
knee=KneeLocator(range(2,21),ss,curve="convex",direction="decreasing")
k=knee.knee
print(k)
    
```

```python
kmc=KMeans(n_clusters=2)
label=kmc.fit_predict(pcaData)
```

```python
sns.scatterplot(x=pcaData[:,0],y=pcaData[:,1],c=label)
```

```python
sns.scatterplot(x=pcaData[:,0],y=pcaData[:,1],c=label)
sns.scatterplot(x=kmc.cluster_centers_[:,0],y=kmc.cluster_centers_[:,1],marker="x",s=150,c="red")
```
